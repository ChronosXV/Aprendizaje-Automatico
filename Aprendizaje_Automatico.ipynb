**Sueño y somnolencia**

El flag datoscompletoSUENO==1 tiene cero faltantes en las variables de sueño. Pero difonsetsueño (diferencia de horario de inicio de sueño entre semana y finde) llega a un máximo de 17,17 horas, que es un valor extraño para una diferencia de horario — más de medio día de diferencia no es creíble

```python
criterio = (
    (df['datoscompletoDEMOG'] == 1) &
    (df['datoscompletoSUENO'] == 1) &
    (df['datoscompletoPANTALLAS'] == 1) &
    (df['excluFINAL'] == 1) &
    (df['edad'] <= 18)
)
d = df[criterio].copy()

print('Distribución de difonsetsueño en los extremos:')
print(d['difonsetsueño'].describe(percentiles=[.9,.95,.99]).round(2))
print()
print('Casos con difonsetsueño > 8h:', (d['difonsetsueño'] > 8).sum())

# Comparar contra el cálculo directo desde las horas de acostarse
cols_horas = ['horasueñosemanahr', 'horaacostarfsemanahr']
print()
print(d[cols_horas].describe().round(2))
```

Esto confirma que hay un grupo de casos calculados mal, las horas base (horasueñosemanahr media 23,15 = ~23:09, horaacostarfsemanahr media 25,17 = ~1:10 AM) ya vienen en formato extendido (pasan de 24 para indicar que cruzan la medianoche), así que la diferencia directa entre esas dos columnas debería alcanzar sin necesidad de ningún ajuste circular.

```python
criterio = (
    (df['datoscompletoDEMOG'] == 1) &
    (df['datoscompletoSUENO'] == 1) &
    (df['datoscompletoPANTALLAS'] == 1) &
    (df['excluFINAL'] == 1) &
    (df['edad'] <= 18)
)
d = df[criterio].copy()

# Diferencia recalculada directamente (ambas columnas ya están en formato extendido >24h)
d['diff_recalculada'] = (d['horaacostarfsemanahr'] - d['horasueñosemanahr']).abs()

print(d['diff_recalculada'].describe(percentiles=[.9,.95,.99]).round(2))
print()

# Comparar contra la columna original del dataset
comparacion = (d['diff_recalculada'] - d['difonsetsueño']).abs()
print('Diferencia entre mi cálculo y la columna original:')
print(comparacion.describe().round(2))
print()
print('Casos donde coinciden (diferencia < 0.1h):', (comparacion < 0.1).sum())
print('Casos donde NO coinciden:', (comparacion >= 0.1).sum())
```

diff_recalculada es mucho más creíble (máximo 9,5h, sin el salto raro del percentil 95), y en 1.596 de 1.728 casos (92,4%) coincide exactamente con la columna original del dataset — eso valida que el cálculo directo es correcto. Los 132 casos donde no coinciden son justamente los que tenían el problema. **Decisión** : descartar difonsetsueño tal como viene en el dataset y usar diff_recalculada (a partir de horasueñosemanahr y horaacostarfsemanahr), debido a la inconsistencia en la variable original en el 7,6% de los casos.

**Relación entre pantallas y sueño/somnolencia**

```python
criterio = (
    (df['datoscompletoDEMOG'] == 1) &
    (df['datoscompletoSUENO'] == 1) &
    (df['datoscompletoPANTALLAS'] == 1) &
    (df['excluFINAL'] == 1) &
    (df['edad'] <= 18)
)
d = df[criterio].copy()

cols_pantalla = ['tpovideojuegoshstrimmed', 'tpojuegotabletcelhstrimmed', 'tporedeshstrimmed', 'tpotvonlinehstrimeed', 'tpootrashstrimed']
d['suma_componentes'] = d[cols_pantalla].sum(axis=1, skipna=False)
d['pantalla_total_final'] = d['tpopantallatotalhstrimmed']
usar_suma = d['pantalla_total_final'].isna() & (d['suma_componentes'] <= 16)
d.loc[usar_suma, 'pantalla_total_final'] = d.loc[usar_suma, 'suma_componentes']

cols_numericas = ['pantalla_total_final'] + cols_pantalla + ['tposueñosem', 'tpofsueñosem', 'Resultado']
print(d[cols_numericas].corr(method='spearman').round(2)['pantalla_total_final'])

print()
print('=== Sueño en semana según nivel de somnolencia ===')
print(d.groupby('somno01pdss')['tposueñosem'].describe()[['count','mean','std']].round(2))
```

El hecho de que el tiempo total de pantalla, tomado como un solo número, no prediga bien sueño ni somnolencia confirma  el argumento que se plantea en el informe: "si todos los adolescentes son tratados como un único grupo, se pierde información".El enfoque agregado no alcanza y por eso se justifica buscar subgrupos con composiciones de uso distintas (no solo "cuánto" sino "en qué" usan la pantalla). El clustering va a permitir caracterizar perfiles distintos, sin garantizar que esos perfiles se diferencien fuertemente en sueño

## Conclusiones del Análisis Exploratorio de Datos (EDA)

**Fuente y estructura del archivo**

El archivo `repositorio.xlsx` contiene 4.115 registros y 125 columnas, sin filas ni columnas duplicadas. El estudio original documenta 1.257 adolescentes, por lo que el archivo disponible es una versión ampliada del dataset publicado.

**Definición de la muestra válida**

El dataset incluye indicadores propios de completitud (`datoscompletoDEMOG`, `datoscompletoSUENO`, `datoscompletoPANTALLAS`, `excluFINAL`). Se verificó que `excluFINAL == 1` (3.251 casos) no garantiza por sí solo la completitud de las variables demográficas: dentro de ese subconjunto, `edad` sigue faltando en 399 casos. Por lo tanto, se definió un criterio propio de inclusión: los tres indicadores de completitud en 1, `excluFINAL == 1` y edad dentro del rango documentado por el estudio (≤18 años). Esto deja una muestra final de **1.728 adolescentes**.

Se evaluó el sesgo de selección comparando el grupo incluido contra el excluido en variables clave: no se encontraron diferencias relevantes en edad (15,44 vs. 15,75 años) ni en tiempo total de pantalla (6,67 vs. 6,75 h). Sí se observó una diferencia en la composición por sexo: el grupo incluido queda balanceado 50/50, mientras que el excluido está sesgado hacia el sexo femenino (59,3%), lo que sugiere que las respuestas incompletas se concentran algo más en ese grupo.

**Tratamiento de outliers y variables reconstruidas**

- *Tiempo total de pantalla:* la variable `tpopantallatotalhstrimmed` tiene un 27,7% de faltantes, producto del tratamiento de valores extremos aplicado por los autores (los outliers se convierten en NaN en lugar de recortarse a un valor). Se verificó que el total es exactamente igual a la suma de sus 5 componentes en todos los casos donde ambos existen, por lo que se reconstruyó una variable `pantalla_total_final` completando los faltantes con la suma de componentes, pero solo cuando esa suma es ≤16 horas (límite de plausibilidad, validado contra el hecho de que el total original nunca supera ese valor). Esto redujo los faltantes de 347 a 202 casos.
- *Diferencia de horario de inicio de sueño:* la variable `difonsetsueño` presentaba valores extremos poco creíbles (hasta 17,17 h de diferencia). Se recalculó directamente como la diferencia absoluta entre `horaacostarfsemanahr` y `horasueñosemanahr`, obteniendo una distribución consistente (máximo 9,5 h) que coincide con la variable original en el 92,4% de los casos, confirmando un error de cálculo en el 7,6% restante.

**Variables descartadas**

Se identificaron variables con niveles de faltantes que las vuelven inviables para el análisis: `WAZ` (99,3% de faltantes) y las variables de rendimiento académico `prom`, `aplazos`, `matematica` y `lengua` (~69% de faltantes cada una).

**Relación entre uso de pantallas y sueño/somnolencia**

Las correlaciones (Spearman) entre el tiempo total de pantalla y las variables de sueño son débiles: -0,11 con horas de sueño en semana, -0,02 con horas de sueño en fin de semana, y 0,09 con el puntaje de somnolencia (PDSS). La diferencia en horas de sueño entre adolescentes con y sin somnolencia positiva es de apenas 19 minutos (7,06 h vs. 7,38 h). Esta ausencia de relación lineal fuerte entre el tiempo *total* de pantalla y el sueño respalda empíricamente el enfoque del trabajo: un análisis agregado no captura diferencias relevantes, lo que justifica buscar perfiles diferenciados según la composición del uso de pantallas mediante clustering, en lugar de predecir a partir de un único indicador. Cabe aclarar que la correlación de Spearman solo captura relaciones monótonas entre el tiempo total de pantalla y el sueño; no descarta que existan patrones diferenciados según la combinación de actividades (redes, videojuegos, TV), que es justamente lo que el enfoque de clustering busca explorar.
