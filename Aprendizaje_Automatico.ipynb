### Búsqueda de duplicados en columnas clave

Investigamos si existen duplicados basándonos en un subconjunto de columnas que deberían identificar entradas únicas, como 'IP' y 'Submission ID'.

```python
key_columns = ['IP', 'Submission ID']
duplicates_in_keys = df[df.duplicated(subset=key_columns, keep=False)]

print(f"Cantidad de filas duplicadas en '{', '.join(key_columns)}': {len(duplicates_in_keys)}")

if not duplicates_in_keys.empty:
    print("Primeras filas duplicadas:")
    display(duplicates_in_keys.sort_values(by=key_columns).head())
else:
    print(f"No se encontraron duplicados en las columnas {', '.join(key_columns)}.")
```
